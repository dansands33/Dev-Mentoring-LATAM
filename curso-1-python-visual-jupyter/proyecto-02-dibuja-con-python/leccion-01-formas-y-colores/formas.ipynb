# Lección 01 — Formas y Colores

Vamos a usar matplotlib como si fuera un lienzo de dibujo.

El sistema de coordenadas funciona así:
- `x` va de izquierda (0) a derecha (1)
- `y` va de abajo (0) a arriba (1)

El punto `(0.5, 0.5)` está justo en el centro.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

%matplotlib inline
print("✅ Matplotlib listo.")
```

## El sistema de coordenadas

Primero entendamos dónde está cada punto.

```python
fig, ax = plt.subplots(figsize=(5, 5))

# Marcamos las esquinas y el centro
puntos_x = [0,   1,   0,   1,   0.5]
puntos_y = [0,   0,   1,   1,   0.5]
nombres  = ["(0,0)", "(1,0)", "(0,1)", "(1,1)", "Centro (0.5, 0.5)"]

ax.scatter(puntos_x, puntos_y, s=100, color="tomato", zorder=5)

for x, y, nombre in zip(puntos_x, puntos_y, nombres):
    ax.annotate(nombre, (x, y), textcoords="offset points", xytext=(8, 5))

ax.set_xlim(-0.1, 1.1)
ax.set_ylim(-0.1, 1.1)
ax.grid(True, alpha=0.3)
ax.set_title("Sistema de coordenadas")
plt.show()
```

## Dibujando una línea

```python
fig, ax = plt.subplots(figsize=(5, 5))

# Línea: de (x_inicio, y_inicio) a (x_fin, y_fin)
ax.plot([0.1, 0.9], [0.2, 0.8], color="steelblue", linewidth=3)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")
ax.set_title("Una línea")
plt.show()
```

## Dibujando un círculo

```python
fig, ax = plt.subplots(figsize=(5, 5))

# Círculo: centro (x, y) y radio
circulo = plt.Circle((0.5, 0.5), 0.3, color="gold", ec="black", linewidth=2)
ax.add_patch(circulo)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Un círculo")
plt.show()
```

## Dibujando un rectángulo

```python
fig, ax = plt.subplots(figsize=(5, 5))

# Rectángulo: esquina inferior izquierda (x, y), ancho, alto
rect = patches.Rectangle((0.2, 0.3), 0.6, 0.4, color="lightcoral", ec="black")
ax.add_patch(rect)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")
ax.set_title("Un rectángulo")
plt.show()
```

## Todo junto en un lienzo

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")
ax.set_facecolor("#1a1a2e")
fig.patch.set_facecolor("#1a1a2e")

# Sol (círculo amarillo)
sol = plt.Circle((0.75, 0.80), 0.12, color="gold")
ax.add_patch(sol)

# Suelo (rectángulo verde)
suelo = patches.Rectangle((0, 0), 1, 0.2, color="#2ecc71")
ax.add_patch(suelo)

# Casa (rectángulo marrón)
casa = patches.Rectangle((0.3, 0.2), 0.4, 0.35, color="#e67e22")
ax.add_patch(casa)

# Techo (triángulo)
from matplotlib.patches import Polygon
techo = Polygon([[0.25, 0.55], [0.75, 0.55], [0.5, 0.75]], color="#c0392b")
ax.add_patch(techo)

# Puerta
puerta = patches.Rectangle((0.44, 0.2), 0.12, 0.2, color="#7f4e1e")
ax.add_patch(puerta)

ax.set_title("Una escena dibujada con formas", color="white")
plt.show()
```
