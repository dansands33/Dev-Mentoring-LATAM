# Lección 03 — Funciones que Dibujan

Una función es un bloque de código con nombre que puedes llamar muchas veces.

```python
def saludar(nombre):
    print(f"Hola, {nombre}!")

saludar("Ana")    # → Hola, Ana!
saludar("Carlos") # → Hola, Carlos!
```

Hoy haremos lo mismo, pero en vez de imprimir texto, vamos a dibujar.

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

%matplotlib inline
print("✅ Listo.")
````

## Función básica de dibujo

````python
def dibujar_punto(ax, x, y, color="blue", radio=0.03):
    """Dibuja un punto en la posición (x, y)."""
    circulo = plt.Circle((x, y), radio, color=color)
    ax.add_patch(circulo)

# Probamos la función
fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")

dibujar_punto(ax, 0.3, 0.5, color="red")
dibujar_punto(ax, 0.5, 0.5, color="blue")
dibujar_punto(ax, 0.7, 0.5, color="green")

ax.set_title("Tres puntos de colores")
plt.show()
````

## Función para dibujar una flor

````python
import numpy as np

def dibujar_flor(ax, cx, cy, radio=0.1, color="pink", n_petalos=6):
    """Dibuja una flor en (cx, cy)."""
    # Petalos
    for i in range(n_petalos):
        angulo = (2 * np.pi / n_petalos) * i
        px = cx + radio * np.cos(angulo)
        py = cy + radio * np.sin(angulo)
        petalo = plt.Circle((px, py), radio * 0.6, color=color, alpha=0.7)
        ax.add_patch(petalo)
    # Centro
    centro = plt.Circle((cx, cy), radio * 0.4, color="gold")
    ax.add_patch(centro)


fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")
ax.set_facecolor("#eaf4ea")
fig.patch.set_facecolor("#eaf4ea")

# Llamamos la función varias veces con distintos valores
dibujar_flor(ax, 0.3, 0.7, radio=0.12, color="violet")
dibujar_flor(ax, 0.6, 0.4, radio=0.08, color="tomato", n_petalos=8)
dibujar_flor(ax, 0.75, 0.75, radio=0.06, color="orange", n_petalos=5)
dibujar_flor(ax, 0.2, 0.3, radio=0.10, color="skyblue")

ax.set_title("Jardín de flores")
plt.show()
````

## Checkpoint ✅

¿Cuántas veces escribiste el código de la flor? Solo una, en `def dibujar_flor(...)`. Eso es el poder de las funciones.

## Función para dibujar una casa

````python
from matplotlib.patches import Polygon

def dibujar_casa(ax, x, y, ancho=0.15, color_pared="orange", color_techo="brown"):
    """Dibuja una casa pequeña con origen en (x, y)."""
    alto = ancho * 0.7
    # Pared
    pared = patches.Rectangle((x, y), ancho, alto, color=color_pared)
    ax.add_patch(pared)
    # Techo
    techo = Polygon([
        [x - 0.01, y + alto],
        [x + ancho + 0.01, y + alto],
        [x + ancho / 2, y + alto + ancho * 0.5]
    ], color=color_techo)
    ax.add_patch(techo)
    # Puerta
    puerta = patches.Rectangle(
        (x + ancho * 0.38, y), ancho * 0.24, alto * 0.5, color="#5d3317"
    )
    ax.add_patch(puerta)


fig, ax = plt.subplots(figsize=(8, 4))
ax.set_xlim(0, 1)
ax.set_ylim(0, 0.6)
ax.axis("off")
ax.set_facecolor("#87CEEB")
fig.patch.set_facecolor("#87CEEB")

# Suelo
suelo = patches.Rectangle((0, 0), 1, 0.05, color="#5d8233")
ax.add_patch(suelo)

# Cuatro casas en una fila
for i in range(4):
    x_casa = 0.05 + i * 0.24
    dibujar_casa(ax, x_casa, 0.05, ancho=0.18)

ax.set_title("Un vecindario")
plt.show()
````
