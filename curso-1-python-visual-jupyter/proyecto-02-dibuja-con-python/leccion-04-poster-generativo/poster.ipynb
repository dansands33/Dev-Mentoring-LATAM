# Lección 04 — Póster Generativo

Hoy vamos a crear una imagen completa generada por código.

Vamos a combinar todo: bucles, funciones, variables y algo de azar.

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import random

%matplotlib inline
print("✅ Todo listo.")
````

## Configuración del póster

Cambia estas variables para personalizar tu póster.

````python
# === PERSONALIZA AQUÍ ===
TITULO = "Mi Póster Generativo"
COLOR_FONDO = "#1a1a2e"
COLORES_ELEMENTOS = ["#e94560", "#0f3460", "#533483", "#e8d5b7", "#06d6a0"]
SEMILLA = 42          # cambia para obtener un diseño diferente
N_CIRCULOS = 30
N_LINEAS = 20
# ========================

random.seed(SEMILLA)
np.random.seed(SEMILLA)
print(f"Configuración cargada. Semilla: {SEMILLA}")
````

## Funciones del póster

````python
def agregar_circulo_decorativo(ax, colores):
    """Agrega un círculo en posición aleatoria."""
    x = random.uniform(0.05, 0.95)
    y = random.uniform(0.15, 0.90)
    radio = random.uniform(0.02, 0.10)
    color = random.choice(colores)
    alpha = random.uniform(0.3, 0.9)
    circulo = plt.Circle((x, y), radio, color=color, alpha=alpha)
    ax.add_patch(circulo)

def agregar_linea_decorativa(ax, colores):
    """Agrega una línea en posición aleatoria."""
    x1, y1 = random.uniform(0, 1), random.uniform(0.15, 0.90)
    x2, y2 = random.uniform(0, 1), random.uniform(0.15, 0.90)
    color = random.choice(colores)
    alpha = random.uniform(0.2, 0.7)
    grosor = random.uniform(0.5, 2.5)
    ax.plot([x1, x2], [y1, y2], color=color, alpha=alpha, linewidth=grosor)

print("Funciones definidas.")
````

## ¡Generamos el póster!

````python
fig, ax = plt.subplots(figsize=(8, 10))

# Fondo
ax.set_facecolor(COLOR_FONDO)
fig.patch.set_facecolor(COLOR_FONDO)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")

# Líneas de fondo
for _ in range(N_LINEAS):
    agregar_linea_decorativa(ax, COLORES_ELEMENTOS)

# Círculos
for _ in range(N_CIRCULOS):
    agregar_circulo_decorativo(ax, COLORES_ELEMENTOS)

# Barra de título
barra = patches.Rectangle((0, 0), 1, 0.12, color="#0f3460", alpha=0.9)
ax.add_patch(barra)

# Texto del título
ax.text(0.5, 0.06, TITULO,
        ha="center", va="center",
        fontsize=18, fontweight="bold",
        color="white")

# Firma
ax.text(0.97, 0.95, f"semilla={SEMILLA}",
        ha="right", va="top",
        fontsize=8, color="white", alpha=0.4)

plt.tight_layout()
plt.show()
````
