# Lección 02 — Patrones con Bucles

Un bucle `for` hace lo mismo muchas veces. Eso es perfecto para dibujar patrones.

```python
for i in range(5):
    print(i)
```

Esto imprime: 0, 1, 2, 3, 4. La variable `i` cambia en cada vuelta.

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

%matplotlib inline

# Primero veamos cómo funciona range
for i in range(6):
    print(f"i = {i}")
````

## Parte 1: Una fila de círculos

Usamos `i` para calcular la posición horizontal de cada círculo.

````python
# Prueba cambiar este número
cantidad = 8

fig, ax = plt.subplots(figsize=(10, 2))
ax.set_xlim(0, cantidad)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")

for i in range(cantidad):
    x = i + 0.5   # posición horizontal: 0.5, 1.5, 2.5...
    y = 0.5       # siempre en el centro vertical
    circulo = plt.Circle((x, y), 0.4, color="steelblue")
    ax.add_patch(circulo)

ax.set_title(f"Fila de {cantidad} círculos")
plt.show()
````

## Parte 2: Círculos con colores que cambian

Usamos `i` para cambiar el color también.

````python
cantidad = 10

fig, ax = plt.subplots(figsize=(12, 2))
ax.set_xlim(0, cantidad)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")

for i in range(cantidad):
    # Calculamos un tono de azul que va cambiando
    intensidad = i / cantidad   # va de 0.0 a 0.9
    color = (intensidad, 0.2, 1 - intensidad)  # (rojo, verde, azul)
    
    circulo = plt.Circle((i + 0.5, 0.5), 0.4, color=color)
    ax.add_patch(circulo)

ax.set_title("Círculos con gradiente de color")
plt.show()
````

## Parte 3: Una cuadrícula

Dos bucles `for` anidados dibujan filas y columnas.

````python
filas = 5
columnas = 8

fig, ax = plt.subplots(figsize=(10, 7))
ax.set_xlim(0, columnas)
ax.set_ylim(0, filas)
ax.set_aspect("equal")
ax.axis("off")

for fila in range(filas):
    for col in range(columnas):
        # El color depende de la posición
        r = col / columnas
        g = fila / filas
        b = 0.5
        
        rect = patches.Rectangle((col, fila), 1, 1, color=(r, g, b))
        ax.add_patch(rect)

ax.set_title("Cuadrícula de colores")
plt.show()
````

## Parte 4: Círculos que crecen

````python
cantidad = 10

fig, ax = plt.subplots(figsize=(8, 3))
ax.set_xlim(0, cantidad)
ax.set_ylim(0, 1.5)
ax.set_aspect("equal")
ax.axis("off")
ax.set_facecolor("#1a1a2e")
fig.patch.set_facecolor("#1a1a2e")

for i in range(cantidad):
    radio = (i + 1) * 0.045   # el radio crece con i
    x = i + 0.5
    circulo = plt.Circle((x, 0.75), radio, color="gold", alpha=0.8)
    ax.add_patch(circulo)

ax.set_title("Círculos que crecen", color="white")
plt.show()
````
