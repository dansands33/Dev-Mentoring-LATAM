# Lección 01 — Primer Gráfico

Vamos a cargar datos reales y hacer un gráfico en menos de 10 líneas de código.

```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
print("✅ Librerías listas.")
```

## Paso 1: Cargar los datos

```python
# Cargamos el archivo CSV
# La ruta es relativa a esta carpeta
ruta = "../../../shared/data/videojuegos.csv"
df = pd.read_csv(ruta)

print(f"Cargados {len(df)} juegos.")
print(f"Columnas: {list(df.columns)}")
```

## Paso 2: Ver la tabla

```python
# Ver las primeras 5 filas
df.head()
```

```python
# Ver estadísticas básicas de las columnas numéricas
df.describe()
```

## Checkpoint ✅

¿Cuál es la calificación máxima? ¿Y la mínima? ¿Cuántas horas promedio tiene el dataset completo?

## Paso 3: Acceder a una columna

```python
# Accedemos a la columna de calificaciones
calificaciones = df["calificacion"]
print(calificaciones)
```

## Paso 4: Hacer el gráfico

```python
# Ordenamos por calificación para que el gráfico se vea mejor
df_ordenado = df.sort_values("calificacion", ascending=False)

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(df_ordenado["titulo"], df_ordenado["calificacion"], color="steelblue")

ax.set_xlabel("Videojuego")
ax.set_ylabel("Calificación")
ax.set_title("Calificaciones de videojuegos populares")
ax.set_ylim(7, 10)   # el eje Y empieza en 7 para que se vean las diferencias

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Paso 5: Gráfico horizontal (más fácil de leer)

```python
fig, ax = plt.subplots(figsize=(8, 10))
ax.barh(df_ordenado["titulo"], df_ordenado["calificacion"], color="tomato")

ax.set_xlabel("Calificación")
ax.set_title("Calificaciones de videojuegos (de mayor a menor)")
ax.set_xlim(7, 10)

plt.tight_layout()
plt.show()
```
