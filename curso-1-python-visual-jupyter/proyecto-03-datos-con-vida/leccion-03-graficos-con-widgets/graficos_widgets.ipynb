# Lección 03 — Gráficos con Widgets

Aquí vamos a conectar widgets con gráficos de datos.

Al cambiar el dropdown, el gráfico cambia solo.

```python
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

%matplotlib inline

df = pd.read_csv("../../../shared/data/videojuegos.csv")
print(f"Dataset listo: {len(df)} juegos")
```

## Parte 1: Filtrar por género con dropdown

```python
generos = ["Todos"] + sorted(df["genero"].unique().tolist())

def graficar_por_genero(genero):
    if genero == "Todos":
        datos = df
    else:
        datos = df[df["genero"] == genero]
    
    datos = datos.sort_values("calificacion", ascending=True)
    
    fig, ax = plt.subplots(figsize=(8, max(3, len(datos) * 0.4)))
    ax.barh(datos["titulo"], datos["calificacion"], color="steelblue")
    ax.set_title(f"Calificaciones — Género: {genero}")
    ax.set_xlabel("Calificación")
    ax.set_xlim(7, 10)
    plt.tight_layout()
    plt.show()

interact(
    graficar_por_genero,
    genero=widgets.Dropdown(options=generos, value="Todos", description="Género:")
)
```

## Parte 2: Slider de año mínimo

```python
año_min = int(df["año"].min())
año_max = int(df["año"].max())

def graficar_desde_año(año_desde):
    filtrado = df[df["año"] >= año_desde].sort_values("calificacion", ascending=False)
    
    fig, ax = plt.subplots(figsize=(10, 5))
    colores = plt.cm.viridis([i / len(filtrado) for i in range(len(filtrado))])
    ax.bar(filtrado["titulo"], filtrado["calificacion"], color=colores)
    ax.set_title(f"Juegos lanzados desde {año_desde} ({len(filtrado)} juegos)")
    ax.set_ylabel("Calificación")
    ax.set_ylim(7, 10)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

interact(
    graficar_desde_año,
    año_desde=widgets.IntSlider(
        min=año_min, max=año_max, step=1, value=año_min,
        description="Desde año:"
    )
)
```

## Parte 3: Comparar dos géneros

```python
generos_lista = sorted(df["genero"].unique().tolist())

def comparar_generos(genero_1, genero_2):
    datos1 = df[df["genero"] == genero_1]
    datos2 = df[df["genero"] == genero_2]
    
    prom1 = datos1["calificacion"].mean()
    prom2 = datos2["calificacion"].mean()
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    for ax, datos, genero, prom in [
        (axes[0], datos1, genero_1, prom1),
        (axes[1], datos2, genero_2, prom2)
    ]:
        ax.bar(datos["titulo"], datos["calificacion"], color="coral")
        ax.axhline(prom, color="navy", linestyle="--", label=f"Promedio: {prom:.2f}")
        ax.set_title(genero)
        ax.set_ylim(7, 10)
        ax.legend()
        ax.tick_params(axis="x", rotation=45)
    
    plt.suptitle(f"{genero_1} vs {genero_2}", fontsize=14)
    plt.tight_layout()
    plt.show()

interact(
    comparar_generos,
    genero_1=widgets.Dropdown(options=generos_lista, value="Acción", description="Género 1:"),
    genero_2=widgets.Dropdown(options=generos_lista, value="RPG", description="Género 2:")
)
```
