# Lección 02 — Explora Datos con Filtros

Un filtro en pandas funciona así:

```python
df[df["calificacion"] > 9]
```

Esto devuelve solo las filas donde la calificación es mayor que 9.

Vamos a hacer preguntas reales a los datos.

````python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

df = pd.read_csv("../../../shared/data/videojuegos.csv")
print(f"Dataset cargado: {len(df)} juegos")
df.head(3)
````

## Pregunta 1: ¿Qué juegos tienen calificación mayor a 9.5?

````python
# Filtramos los juegos con calificación > 9.5
top = df[df["calificacion"] > 9.5]

print(f"Juegos con calificación > 9.5: {len(top)}")
top[["titulo", "calificacion", "genero"]]
````

## Pregunta 2: ¿Cuántos juegos hay de cada género?

````python
conteo = df["genero"].value_counts()
print(conteo)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(conteo.index, conteo.values, color="mediumpurple")
ax.set_title("Juegos por género")
ax.set_xlabel("Género")
ax.set_ylabel("Cantidad")
plt.tight_layout()
plt.show()
````

## Pregunta 3: ¿Cuál es la calificación promedio por género?

````python
promedio_por_genero = df.groupby("genero")["calificacion"].mean().sort_values(ascending=False)
print(promedio_por_genero)

fig, ax = plt.subplots(figsize=(8, 4))
ax.barh(promedio_por_genero.index, promedio_por_genero.values, color="teal")
ax.set_title("Calificación promedio por género")
ax.set_xlabel("Calificación promedio")
ax.set_xlim(7, 10)
plt.tight_layout()
plt.show()
````

## Pregunta 4: ¿Los juegos más recientes tienen mejor calificación?

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df["año"], df["calificacion"], s=80, color="coral", alpha=0.7)

# Agregamos etiquetas a los puntos extremos
for _, fila in df.iterrows():
    if fila["calificacion"] >= 9.6 or fila["calificacion"] <= 7.9:
        ax.annotate(fila["titulo"], (fila["año"], fila["calificacion"]),
                   textcoords="offset points", xytext=(5, 3), fontsize=8)

ax.set_title("Año vs Calificación")
ax.set_xlabel("Año de lanzamiento")
ax.set_ylabel("Calificación")
plt.tight_layout()
plt.show()
````

## Pregunta 5: ¿Qué juego tiene más horas de juego?

````python
# El juego con más horas
max_horas = df.loc[df["horas_promedio"].idxmax()]
print(f"Juego con más horas: {max_horas['titulo']} ({max_horas['horas_promedio']} horas)")

# El juego con menos horas
min_horas = df.loc[df["horas_promedio"].idxmin()]
print(f"Juego con menos horas: {min_horas['titulo']} ({min_horas['horas_promedio']} horas)")
````
