```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact
from IPython.display import display

# Configurar matplotlib para modo interactivo
%matplotlib inline

print("✅ Widgets listos.")
```

## Parte 1: Slider que controla el tamaño

Mueve el slider y mira cómo cambia el círculo.

```python
def dibujar_circulo(tamaño):
    fig, ax = plt.subplots(figsize=(4, 4))
    circulo = plt.Circle((0.5, 0.5), tamaño / 100, color="steelblue")
    ax.add_patch(circulo)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect("equal")
    ax.axis("off")
    ax.set_title(f"Tamaño: {tamaño}%")
    plt.show()

interact(dibujar_circulo, tamaño=widgets.IntSlider(min=5, max=45, step=1, value=20))
```

## Parte 2: Slider que controla la cantidad

```python
def dibujar_puntos(cantidad):
    import numpy as np
    np.random.seed(42)  # semilla fija para posiciones consistentes
    
    fig, ax = plt.subplots(figsize=(5, 5))
    x = np.random.rand(cantidad)
    y = np.random.rand(cantidad)
    ax.scatter(x, y, s=100, color="tomato", alpha=0.7)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_title(f"{cantidad} puntos")
    ax.axis("off")
    plt.show()

interact(dibujar_puntos, cantidad=widgets.IntSlider(min=1, max=50, step=1, value=10))
```

## Checkpoint ✅

¿Viste cómo los puntos aparecen y desaparecen al mover el slider? Eso es interactividad real.

## Parte 3: Selector de color

```python
def mostrar_color(color):
    import matplotlib.patches as patches
    fig, ax = plt.subplots(figsize=(5, 2))
    rect = patches.Rectangle((0.05, 0.1), 0.9, 0.8, color=color)
    ax.add_patch(rect)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")
    ax.set_title(f"Color seleccionado: {color}")
    plt.show()

interact(
    mostrar_color,
    color=widgets.ColorPicker(value="#3498db", description="Elige un color:")
)
```

## Parte 4: Botón que hace algo

Al hacer clic en el botón, el color cambia por uno aleatorio.

```python
import random
import matplotlib.patches as patches

output = widgets.Output()
boton = widgets.Button(description="¡Cambiar color!", button_style="success")

colores = ["red", "blue", "green", "orange", "purple", "pink", "cyan", "gold"]

def al_hacer_clic(b):
    color = random.choice(colores)
    with output:
        output.clear_output(wait=True)
        fig, ax = plt.subplots(figsize=(4, 2))
        rect = patches.Rectangle((0.05, 0.1), 0.9, 0.8, color=color)
        ax.add_patch(rect)
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.axis("off")
        ax.set_title(f"Color: {color}")
        plt.show()

boton.on_click(al_hacer_clic)
display(boton, output)
```
