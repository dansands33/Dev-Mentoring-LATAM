# Lección 02 — Variables que Cambian el Mundo

En esta lección vas a ver que **cambiar una variable cambia lo que aparece en pantalla**.

Eso es poder real: tú controlas lo que Python muestra.

```python
# Importamos matplotlib para dibujar
import matplotlib.pyplot as plt
import matplotlib.patches as patches

print("✅ Listo. Ahora podemos dibujar.")
```

## Parte 1: Una variable controla el tamaño

Cambia el valor de `tamaño` y ejecuta la celda. ¿Qué pasa con el círculo?

```python
# Cambia este número (prueba con 0.1, 0.5, 0.9)
tamaño = 0.3

fig, ax = plt.subplots(figsize=(4, 4))
circulo = plt.Circle((0.5, 0.5), tamaño, color="steelblue")
ax.add_patch(circulo)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title(f"Tamaño: {tamaño}")
plt.show()
```

## Checkpoint ✅

Cambia `tamaño` a `0.05` y a `0.45`. ¿El círculo cambia?

## Parte 2: Una variable controla el color

Prueba con: `"red"`, `"green"`, `"orange"`, `"purple"`, `"#FF5733"`

```python
# Cambia el color
color = "steelblue"

fig, ax = plt.subplots(figsize=(4, 2))
rect = patches.Rectangle((0.1, 0.2), 0.8, 0.6, color=color)
ax.add_patch(rect)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis("off")
ax.set_title(f"Color: {color}")
plt.show()
```

## Parte 3: Una variable controla la cantidad

Cambia `cantidad` y mira cuántos círculos aparecen.

```python
import numpy as np

# Cambia este número (prueba entre 1 y 20)
cantidad = 5
color = "tomato"

fig, ax = plt.subplots(figsize=(8, 2))

for i in range(cantidad):
    x = (i + 0.5) / cantidad  # posición horizontal de cada círculo
    circulo = plt.Circle((x, 0.5), 0.35 / cantidad, color=color)
    ax.add_patch(circulo)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title(f"Cantidad: {cantidad} círculos")
plt.show()
```

## Checkpoint ✅

¿Qué pasa si pones `cantidad = 1`? ¿Y si pones `cantidad = 15`?

Nota cómo el código no cambia, solo la variable.

## Parte 4: Variables de texto también controlan cosas

```python
# Cambia estos textos
titulo = "Mi primer gráfico"
subtitulo = "Hecho con Python"
color_fondo = "lightyellow"

fig, ax = plt.subplots(figsize=(6, 3))
fig.patch.set_facecolor(color_fondo)
ax.set_facecolor(color_fondo)
ax.text(0.5, 0.6, titulo, ha="center", va="center", fontsize=20, fontweight="bold")
ax.text(0.5, 0.35, subtitulo, ha="center", va="center", fontsize=14, color="gray")
ax.axis("off")
plt.show()
```
