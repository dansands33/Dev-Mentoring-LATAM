## Variables: guardar información

Una variable es como una caja con nombre. Guardas algo adentro y puedes usarlo después.

````python
nombre = "Estudiante"  # cambia esto por tu nombre
carrera = "Ingeniería de Software"
año = 2026

print("Hola, soy", nombre)
print("Estudio", carrera)
print("Año:", año)
````

## Checkpoint ✅

¿Viste tu nombre en pantalla? Bien. Eso significa que la variable funcionó.

Ahora prueba cambiar el año y vuelve a ejecutar. ¿Cambió el mensaje?

````python
# También puedes combinar texto con variables usando f-strings
mensaje = f"Hola, soy {nombre} y estudio {carrera}."
print(mensaje)
````

## Variables numéricas

Las variables también pueden guardar números. Y puedes hacer operaciones con ellas.

````python
edad = 20
años_para_graduarse = 4

graduacion = edad + años_para_graduarse

print(f"Tengo {edad} años.")
print(f"Me graduaré a los {graduacion} años.")
````

## Checkpoint ✅

Cambia `edad` a tu edad real y vuelve a ejecutar. ¿Cambió el cálculo?

Eso es exactamente cómo funcionan las variables: guardas un valor, lo usas en cualquier parte.
