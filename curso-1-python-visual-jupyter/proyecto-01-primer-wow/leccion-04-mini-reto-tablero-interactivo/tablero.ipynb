Ejecuta todas las celdas en orden y luego interactúa con los controles.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display

%matplotlib inline
print("✅ Todo importado.")
```

## Los controles del tablero

Aquí definimos todos los controles. Los puedes ver y usar después de ejecutar esta celda.

```python
# Control de figura
selector_figura = widgets.Dropdown(
    options=["círculo", "cuadrado", "triángulo"],
    value="círculo",
    description="Figura:"
)

# Control de color
selector_color = widgets.ColorPicker(
    value="#3498db",
    description="Color:"
)

# Control de tamaño
slider_tamaño = widgets.IntSlider(
    min=10, max=90, step=5, value=40,
    description="Tamaño %:"
)

# Campo de texto para el mensaje
campo_mensaje = widgets.Text(
    value="¡Hola Python!",
    description="Mensaje:",
    layout=widgets.Layout(width="300px")
)

print("Controles creados.")
```

## La función del tablero

Esta función recibe los valores de los controles y dibuja el tablero.

```python
def actualizar_tablero(figura, color, tamaño, mensaje):
    radio = tamaño / 100
    
    fig, ax = plt.subplots(figsize=(6, 5))
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect("equal")
    ax.set_facecolor("#f8f9fa")
    ax.axis("off")

    cx, cy = 0.5, 0.55  # centro del dibujo

    if figura == "círculo":
        forma = plt.Circle((cx, cy), radio / 2, color=color)
        ax.add_patch(forma)

    elif figura == "cuadrado":
        lado = radio
        forma = patches.Rectangle(
            (cx - lado / 2, cy - lado / 2), lado, lado, color=color
        )
        ax.add_patch(forma)

    elif figura == "triángulo":
        import matplotlib.patches as mpatches
        from matplotlib.patches import Polygon
        h = radio * 0.866  # altura del triángulo equilátero
        puntos = [
            [cx, cy + radio * 0.6],
            [cx - radio / 2, cy - radio * 0.4],
            [cx + radio / 2, cy - radio * 0.4]
        ]
        forma = Polygon(puntos, color=color)
        ax.add_patch(forma)

    # Mensaje debajo de la figura
    ax.text(0.5, 0.1, mensaje, ha="center", va="center",
            fontsize=14, fontweight="bold", color="#2c3e50")

    plt.tight_layout()
    plt.show()

print("Función lista.")
```

## ¡Activa el tablero!

Ejecuta esta celda y después interactúa con los controles de arriba.

```python
salida = widgets.interactive_output(
    actualizar_tablero,
    {
        "figura": selector_figura,
        "color": selector_color,
        "tamaño": slider_tamaño,
        "mensaje": campo_mensaje
    }
)

# Organizar controles en una caja
controles = widgets.VBox([
    selector_figura,
    selector_color,
    slider_tamaño,
    campo_mensaje
])

# Mostrar controles y tablero juntos
display(widgets.HBox([controles, salida]))
```
