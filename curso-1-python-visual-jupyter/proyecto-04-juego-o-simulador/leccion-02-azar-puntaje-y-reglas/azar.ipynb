# Lección 02 — Azar, Puntaje y Reglas

El azar hace que los juegos sean diferentes cada vez. En Python usamos `random` para eso.

```python
import random
print("✅ random importado.")
```

## Parte 1: Dado virtual

Ejecuta esta celda varias veces. El resultado cambia cada vez.

```python
# Número aleatorio entre 1 y 6 (como un dado)
resultado = random.randint(1, 6)
print(f"🎲 Sacaste un {resultado}")
```

## Parte 2: Elección aleatoria de una lista

```python
opciones = ["piedra", "papel", "tijeras"]
eleccion_maquina = random.choice(opciones)
print(f"La máquina eligió: {eleccion_maquina}")
```

## Parte 3: Puntaje que se acumula

`puntaje += x` suma `x` al puntaje actual.

```python
puntaje = 0
rondas = 5

print(f"Jugando {rondas} rondas...\n")

for ronda in range(1, rondas + 1):
    dado = random.randint(1, 6)
    
    if dado >= 4:          # ganas con 4, 5 o 6
        puntos = dado      # los puntos son el número del dado
        resultado = "✅ Ganaste"
    else:
        puntos = 0
        resultado = "❌ Perdiste"
    
    puntaje += puntos
    print(f"Ronda {ronda}: dado={dado} → {resultado} (+{puntos} puntos)")

print(f"\nPuntaje final: {puntaje} / {rondas * 6} posibles")
```

## Parte 4: Mini juego de adivinar el número

```python
# El juego genera un número secreto
# El jugador tiene N intentos para acercarse

numero_secreto = random.randint(1, 10)
intentos = [3, 7, 5, numero_secreto]  # simula intentos del jugador

print(f"(El número secreto es {numero_secreto} — en el juego real no lo verías)\n")

for i, intento in enumerate(intentos, 1):
    if intento == numero_secreto:
        print(f"Intento {i}: {intento} → 🎉 ¡Correcto! Lo lograste en {i} intentos.")
        break
    elif intento < numero_secreto:
        print(f"Intento {i}: {intento} → ⬆️ Más alto")
    else:
        print(f"Intento {i}: {intento} → ⬇️ Más bajo")
```
