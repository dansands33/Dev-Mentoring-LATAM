````python
import random
import ipywidgets as widgets
from IPython.display import display

print("✅ Todo importado.")
````

## Estado del juego

````python
estado = {
    "victorias": 0,
    "derrotas": 0,
    "empates": 0,
    "historial": []
}

OPCIONES = ["piedra", "papel", "tijeras"]
EMOJIS = {"piedra": "🪨", "papel": "📄", "tijeras": "✂️"}

print("Estado del juego inicializado.")
````

## Lógica del juego

````python
def determinar_ganador(jugador, maquina):
    """Devuelve 'jugador', 'maquina' o 'empate'."""
    if jugador == maquina:
        return "empate"
    
    # Combinaciones ganadoras para el jugador
    gana_jugador = [
        ("piedra", "tijeras"),
        ("papel", "piedra"),
        ("tijeras", "papel")
    ]
    
    if (jugador, maquina) in gana_jugador:
        return "jugador"
    else:
        return "maquina"

# Prueba rápida
print(determinar_ganador("piedra", "tijeras"))  # debería ser 'jugador'
print(determinar_ganador("papel", "papel"))      # debería ser 'empate'
print(determinar_ganador("piedra", "papel"))     # debería ser 'maquina'
````

## Los widgets

````python
# Tres botones de elección
btn_piedra  = widgets.Button(description="🪨 Piedra",  button_style="primary",
                             layout=widgets.Layout(width="120px", height="50px"))
btn_papel   = widgets.Button(description="📄 Papel",   button_style="info",
                             layout=widgets.Layout(width="120px", height="50px"))
btn_tijeras = widgets.Button(description="✂️ Tijeras", button_style="warning",
                             layout=widgets.Layout(width="120px", height="50px"))
btn_reiniciar = widgets.Button(description="🔄 Reiniciar", button_style="danger",
                               layout=widgets.Layout(width="120px", height="35px"))

# Marcador
label_marcador = widgets.HTML(
    value="<b style='font-size:16px'>🏆 Victorias: 0 | 💀 Derrotas: 0 | 🤝 Empates: 0</b>"
)

# Resultado de la última ronda
output_ronda = widgets.Output()

print("Widgets listos.")
````

## Lógica de los botones

````python
def actualizar_marcador():
    label_marcador.value = (
        f"<b style='font-size:16px'>"
        f"🏆 Victorias: {estado['victorias']} | "
        f"💀 Derrotas: {estado['derrotas']} | "
        f"🤝 Empates: {estado['empates']}"
        f"</b>"
    )


def jugar(eleccion_jugador):
    eleccion_maquina = random.choice(OPCIONES)
    ganador = determinar_ganador(eleccion_jugador, eleccion_maquina)

    emoji_j = EMOJIS[eleccion_jugador]
    emoji_m = EMOJIS[eleccion_maquina]

    if ganador == "jugador":
        estado["victorias"] += 1
        resultado_txt = "🎉 ¡Ganaste!"
    elif ganador == "maquina":
        estado["derrotas"] += 1
        resultado_txt = "😤 Perdiste."
    else:
        estado["empates"] += 1
        resultado_txt = "🤝 Empate."

    linea = f"Tú: {emoji_j}  Máquina: {emoji_m}  → {resultado_txt}"
    estado["historial"].append(linea)
    actualizar_marcador()

    with output_ronda:
        output_ronda.clear_output(wait=True)
        print(linea)
        print()
        # Mostrar últimas 5 jugadas
        if len(estado["historial"]) > 1:
            print("Últimas jugadas:")
            for h in estado["historial"][-5:]:
                print(" ", h)


def reiniciar(b):
    estado["victorias"] = 0
    estado["derrotas"] = 0
    estado["empates"] = 0
    estado["historial"] = []
    actualizar_marcador()
    with output_ronda:
        output_ronda.clear_output()
        print("Juego reiniciado. ¡A jugar!")


# Conectar botones
btn_piedra.on_click(lambda b: jugar("piedra"))
btn_papel.on_click(lambda b: jugar("papel"))
btn_tijeras.on_click(lambda b: jugar("tijeras"))
btn_reiniciar.on_click(reiniciar)

print("Lógica conectada.")
````

## ¡A jugar!

````python
display(
    widgets.VBox([
        widgets.HTML("<h2 style='text-align:center'>🎮 Piedra, Papel o Tijeras</h2>"),
        widgets.HTML("<p style='text-align:center'>Elige tu opción:</p>"),
        widgets.HBox(
            [btn_piedra, btn_papel, btn_tijeras],
            layout=widgets.Layout(justify_content="center")
        ),
        widgets.HTML("<hr>"),
        label_marcador,
        output_ronda,
        widgets.HTML("<hr>"),
        btn_reiniciar
    ])
)
````
