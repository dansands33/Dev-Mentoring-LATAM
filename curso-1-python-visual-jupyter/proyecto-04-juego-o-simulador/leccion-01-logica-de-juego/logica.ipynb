# Lección 01 — Lógica de Juego

Todo juego tiene reglas. Las reglas se escriben con `if`, `elif` y `else`.

```python
if condicion:
    # esto pasa si condicion es True
elif otra_condicion:
    # esto pasa si la primera falló pero esta es True
else:
    # esto pasa si ninguna fue True
```

## Ejemplo 1: ¿Ganaste o perdiste?

````python
# Cambia este número y ejecuta
puntaje = 45
puntaje_para_ganar = 50

if puntaje >= puntaje_para_ganar:
    print("🏆 ¡Ganaste!")
else:
    faltaron = puntaje_para_ganar - puntaje
    print(f"😞 Perdiste. Te faltaron {faltaron} puntos.")
````

## Ejemplo 2: Varios niveles

````python
# Prueba con diferentes valores de puntaje
puntaje = 72

if puntaje >= 90:
    nivel = "S — Maestro"
    emoji = "⭐⭐⭐"
elif puntaje >= 70:
    nivel = "A — Experto"
    emoji = "⭐⭐"
elif puntaje >= 50:
    nivel = "B — Intermedio"
    emoji = "⭐"
elif puntaje >= 30:
    nivel = "C — Básico"
    emoji = "🔵"
else:
    nivel = "D — Principiante"
    emoji = "🔴"

print(f"Puntaje: {puntaje}")
print(f"Nivel: {nivel} {emoji}")
````

## Ejemplo 3: Reglas de un juego de cartas simple

````python
# Simulamos dos cartas (números del 1 al 10)
carta_jugador = 7
carta_maquina = 5

print(f"Tu carta: {carta_jugador}")
print(f"Carta de la máquina: {carta_maquina}")
print()

if carta_jugador > carta_maquina:
    print("🎉 ¡Tú ganas esta ronda!")
elif carta_jugador < carta_maquina:
    print("😤 La máquina gana esta ronda.")
else:
    print("🤝 Empate. Nadie gana puntos.")
````

## Ejemplo 4: Condiciones combinadas con `and` / `or`

````python
vidas = 3
puntaje = 80
tiene_escudo = True

# El jugador sobrevive si tiene vidas Y (puntaje alto O escudo)
if vidas > 0 and (puntaje >= 50 or tiene_escudo):
    print("✅ El jugador sobrevive al nivel.")
else:
    print("💀 Game Over.")

# Prueba cambiando vidas = 0 o tiene_escudo = False
````
