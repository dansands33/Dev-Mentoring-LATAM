# Lección 03 — Interfaz Simple en Notebook

Vamos a crear un juego con botones. El estado del juego (puntaje, rondas) se guarda entre clics.

````python
import random
import ipywidgets as widgets
from IPython.display import display

print("✅ Todo listo.")
````

## Estado del juego

El estado vive fuera de las funciones para que persista entre clics.

````python
# Estado global del juego
estado = {
    "puntaje": 0,
    "rondas": 0,
    "historial": []
}

print("Estado inicial:", estado)
````

## Los widgets del juego

````python
# Botón principal
boton_tirar = widgets.Button(
    description="🎲 Tirar dado",
    button_style="primary",
    layout=widgets.Layout(width="150px", height="40px")
)

# Botón reiniciar
boton_reiniciar = widgets.Button(
    description="🔄 Reiniciar",
    button_style="warning",
    layout=widgets.Layout(width="130px", height="40px")
)

# Panel de puntaje (texto que se actualiza)
label_puntaje = widgets.HTML(value="<b>Puntaje: 0 | Rondas: 0</b>")

# Área de resultados
output_resultado = widgets.Output()

print("Widgets listos.")
````

## Las funciones de los botones

````python
def tirar_dado(b):
    dado = random.randint(1, 6)
    estado["rondas"] += 1

    if dado >= 4:
        puntos = dado
        icono = "✅"
        mensaje = f"Sacaste un {dado}. ¡Ganaste {puntos} puntos!"
    elif dado == 1:
        puntos = -2
        icono = "💀"
        mensaje = f"Sacaste un 1. Pierdes 2 puntos."
    else:
        puntos = 0
        icono = "❌"
        mensaje = f"Sacaste un {dado}. Sin puntos esta ronda."

    estado["puntaje"] += puntos
    estado["historial"].append(f"Ronda {estado['rondas']}: {icono} {mensaje}")

    # Actualizar puntaje
    label_puntaje.value = f"<b>Puntaje: {estado['puntaje']} | Rondas: {estado['rondas']}</b>"

    # Mostrar historial
    with output_resultado:
        output_resultado.clear_output(wait=True)
        for linea in estado["historial"][-5:]:  # últimas 5 rondas
            print(linea)


def reiniciar(b):
    estado["puntaje"] = 0
    estado["rondas"] = 0
    estado["historial"] = []
    label_puntaje.value = "<b>Puntaje: 0 | Rondas: 0</b>"
    with output_resultado:
        output_resultado.clear_output()
        print("Juego reiniciado. ¡Buena suerte!")


boton_tirar.on_click(tirar_dado)
boton_reiniciar.on_click(reiniciar)
print("Funciones conectadas.")
````

## ¡Juega!

````python
display(
    widgets.VBox([
        widgets.HTML("<h3>🎲 Juego de Dados</h3>"),
        label_puntaje,
        widgets.HBox([boton_tirar, boton_reiniciar]),
        output_resultado
    ])
)
````
